````python
import xarray as xr
import numpy as np

# Open your Zarr store if not already loaded
ds = xr.open_zarr("data/glorys_gom_jan2004.zarr")

# Identify SST and SSH variables in your dataset
# Adjust variable names if yours are named differently (e.g. 'thetao' instead of 'tos')
sst_var = "tos" if "tos" in ds else ("thetao" if "thetao" in ds else list(ds.data_vars)[0])
ssh_var = "zos" if "zos" in ds else list(ds.data_vars)[1]

# If SST has a depth dimension, slice the surface (depth=0)
sst_data = ds[sst_var].isel(depth=0) if "depth" in ds[sst_var].dims else ds[sst_var]
ssh_data = ds[ssh_var]

# Compute summary stats (loads into memory fast because it's rechunked!)
stats = {}
for name, da in [("SST", sst_data), ("SSH", ssh_data)]:
    stats[name] = {
        "min": float(da.min().compute()),
        "max": float(da.max().compute()),
        "mean": float(da.mean().compute()),
        "variance": float(da.var().compute()),
        "std_dev": float(da.std().compute()),
    }

# Display results formatted
for var_name, s in stats.items():
    unit = "°C" if var_name == "SST" else "m"
    print(f"\n[{var_name}] ({unit})")
    print(f"  Min:      {s['min']:.2f} {unit}")
    print(f"  Max:      {s['max']:.2f} {unit}")
    print(f"  Mean:     {s['mean']:.2f} {unit}")
    print(f"  Variance: {s['variance']:.4f}")
    print(f"  Std Dev:  {s['std_dev']:.2f} {unit}")
````

````python
import numpy as np
from scipy.spatial.distance import cdist
import pandas as pd

def extract_front_points(sst_grid, threshold_grad=0.05):
    """
    Extracts grid index coordinates (i, j) where spatial SST gradient is high.
    Replaces NaNs with nearest values first to prevent gradient leakage on coasts.
    """
    # Create a clean copy and handle NaNs (land mask)
    grid_clean = np.nan_to_num(sst_grid, nan=np.nanmean(sst_grid))
    
    # Compute spatial gradients across grid dimensions
    dy, dx = np.gradient(grid_clean)
    grad_mag = np.sqrt(dx**2 + dy**2)
    
    # Mask out land from gradient calculation
    grad_mag[np.isnan(sst_grid)] = 0.0
    
    # Extract index pairs (row, col) above gradient threshold
    front_indices = np.where(grad_mag > threshold_grad)
    
    # Stack into an (N, 2) point array
    points = np.column_stack((front_indices[0], front_indices[1]))
    return points

def compute_mhd(set_A, set_B):
    """Computes Modified Hausdorff Distance between two 2D point sets A and B."""
    if len(set_A) == 0 or len(set_B) == 0:
        return np.inf  # Handle empty front edge cases
        
    d_matrix = cdist(set_A, set_B, metric='euclidean')
    d_A_B = np.mean(np.min(d_matrix, axis=1))
    d_B_A = np.mean(np.min(d_matrix, axis=0))
    return max(d_A_B, d_B_A)

# --- Compute MHD across Search Window ---
target_points = extract_front_points(target_sst.values, threshold_grad=0.05)

mhd_scores = {}
for time_step in historical_pool.time.values:
    hist_sst = historical_pool.sel(time=time_step).values
    hist_points = extract_front_points(hist_sst, threshold_grad=0.05)
    
    # Measure structural match
    score = compute_mhd(target_points, hist_points)
    mhd_scores[time_step] = score

# Sort and select Top K analogs via MHD
mhd_series = pd.Series(mhd_scores)
top_k_mhd = mhd_series.sort_values().head(5)

for rank, (t_val, score) in enumerate(top_k_mhd.items(), 1):
    date_str = pd.to_datetime(t_val).strftime("%Y-%m-%d")
    print(f"Rank {rank}: Date = {date_str} | MHD = {score:.4f} grid units")
````

````python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import numpy as np

class SSTDataset(Dataset):
    def __init__(self, sst_xarray, img_size=(64, 64)):
        """
        Extracts 2D SST grids, handles NaNs, and normalizes values to [0, 1].
        """
        raw_data = sst_xarray.values # Shape: (time, lat, lon)
        
        # 1. Fill land NaNs with area average (prevents NaN gradient propagation)
        mean_val = np.nanmean(raw_data)
        clean_data = np.nan_to_num(raw_data, nan=mean_val)
        
        # 2. Min-Max Normalization across dataset
        self.min_val = clean_data.min()
        self.max_val = clean_data.max()
        norm_data = (clean_data - self.min_val) / (self.max_val - self.min_val)
        
        # 3. Convert to PyTorch Tensor: (B, C, H, W)
        tensors = torch.tensor(norm_data, dtype=torch.float32).unsqueeze(1)
        
        # 4. Resize to power of 2 for CNN downsampling
        self.data = F.interpolate(tensors, size=img_size, mode='bilinear', align_corners=False)

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        return self.data[idx]

# Instantiate dataset from your xarray object
dataset = SSTDataset(sst_data, img_size=(64, 64))
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)
````

````python
class ConvVAE(nn.Module):
    def __init__(self, latent_dim=32):
        super(ConvVAE, self).__init__()
        self.latent_dim = latent_dim

        # Encoder: Downsamples (1, 64, 64) -> (128, 8, 8)
        self.encoder = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=4, stride=2, padding=1),  # -> (32, 32, 32)
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=4, stride=2, padding=1), # -> (64, 16, 16)
            nn.ReLU(),
            nn.Conv2d(64, 128, kernel_size=4, stride=2, padding=1),# -> (128, 8, 8)
            nn.ReLU(),
            nn.Flatten()
        )
        
        # Latent projection parameters
        self.fc_mu = nn.Linear(128 * 8 * 8, latent_dim)
        self.fc_logvar = nn.Linear(128 * 8 * 8, latent_dim)

        # Decoder Input
        self.decoder_input = nn.Linear(latent_dim, 128 * 8 * 8)

        # Decoder: Upsamples back to (1, 64, 64)
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(128, 64, kernel_size=4, stride=2, padding=1), # -> (64, 16, 16)
            nn.ReLU(),
            nn.ConvTranspose2d(64, 32, kernel_size=4, stride=2, padding=1),  # -> (32, 32, 32)
            nn.ReLU(),
            nn.ConvTranspose2d(32, 1, kernel_size=4, stride=2, padding=1),   # -> (1, 64, 64)
            nn.Sigmoid() # Keeps outputs bounded between 0 and 1
        )

    def reparameterize(self, mu, logvar):
        """Sampling trick for backpropagation."""
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def encode(self, x):
        h = self.encoder(x)
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        return mu, logvar

    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        reconstruction = self.decoder(self.decoder_input(z).view(-1, 128, 8, 8))
        return reconstruction, mu, logvar
````

````python
import matplotlib.pyplot as plt
from IPython.display import clear_output, display
import torch
import torch.nn.functional as F

# --- Loss Function Definition ---
def vae_loss_function(recon_x, x, mu, logvar, beta=0.001):
    # MSE reconstruction loss
    recon_loss = F.mse_loss(recon_x, x, reduction='sum')
    
    # KL Divergence regularization
    kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    
    total_loss = recon_loss + beta * kl_loss
    return total_loss, recon_loss, kl_loss

# --- Setup Model & Optimizer ---
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
vae = ConvVAE(latent_dim=32).to(device)
optimizer = torch.optim.Adam(vae.parameters(), lr=1e-3)

epochs = 100
vae.train()

# --- Loss Tracking ---
epoch_total_losses = []
epoch_recon_losses = []
epoch_kl_losses = []

# --- Interactive Plot Setup ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))

print("Pretraining VAE on 2004 SST Grids...")

for epoch in range(1, epochs + 1):
    running_total = 0.0
    running_recon = 0.0
    running_kl = 0.0
    
    for batch in dataloader:
        batch = batch.to(device)
        optimizer.zero_grad()
        
        recon_batch, mu, logvar = vae(batch)
        loss, recon_loss, kl_loss = vae_loss_function(recon_batch, batch, mu, logvar, beta=0.001)
        
        loss.backward()
        optimizer.step()
        
        running_total += loss.item()
        running_recon += recon_loss.item()
        running_kl += kl_loss.item()
        
    # Calculate average loss per sample for the epoch
    n_samples = len(dataloader.dataset)
    avg_total = running_total / n_samples
    avg_recon = running_recon / n_samples
    avg_kl = running_kl / n_samples
    
    epoch_total_losses.append(avg_total)
    epoch_recon_losses.append(avg_recon)
    epoch_kl_losses.append(avg_kl)
    
    # --- Live Plot Update ---
    clear_output(wait=True)
    ax1.clear()
    ax2.clear()
    
    # Left subplot: Total VAE Loss
    ax1.plot(range(1, epoch + 1), epoch_total_losses, color='tab:blue', linewidth=2, label='Total Loss')
    ax1.scatter(epoch, avg_total, color='tab:red', s=30, zorder=5)
    ax1.set_title(f"Total Loss (Epoch {epoch}/{epochs}: {avg_total:.2f})")
    ax1.set_xlabel("Epoch")
    ax1.set_ylabel("Loss per Sample")
    ax1.grid(True, linestyle='--', alpha=0.5)
    ax1.legend(loc='upper right')
    
    # Right subplot: Breakdown (Reconstruction vs. KL)
    ax2.plot(range(1, epoch + 1), epoch_recon_losses, color='tab:green', linewidth=2, label='MSE Recon Loss')
    ax2.plot(range(1, epoch + 1), epoch_kl_losses, color='tab:orange', linewidth=2, label='KL Div Loss')
    ax2.set_title(f"Loss Components (Recon: {avg_recon:.2f} | KL: {avg_kl:.2f})")
    ax2.set_xlabel("Epoch")
    ax2.grid(True, linestyle='--', alpha=0.5)
    ax2.legend(loc='upper right')
    
    plt.tight_layout()
    display(fig)

plt.close(fig)  # Prevents duplicate static rendering at the end of execution
print(f"VAE Training Complete! Final Loss: {epoch_total_losses[-1]:.4f}")
````

````python
vae.eval()
latent_embeddings = {}

with torch.no_grad():
    for i, time_step in enumerate(sst_data.time.values):
        # Extract single slice and format
        grid_tensor = dataset[i].unsqueeze(0).to(device)
        mu, _ = vae.encode(grid_tensor)
        latent_embeddings[time_step] = mu.cpu().numpy().flatten()

# Perform Analog Search in Latent Space
target_z = latent_embeddings[np.datetime64('2004-07-01')]

latent_distances = {}
for time_step, hist_z in latent_embeddings.items():
    if time_step != np.datetime64('2004-07-01'):
        # Calculate Euclidean RMSD in latent space
        dist = np.sqrt(np.mean((target_z - hist_z) ** 2))
        latent_distances[time_step] = dist

# Top 5 Latent Analogs
top_5_latent = pd.Series(latent_distances).sort_values().head(5)
print("Top 5 VAE Latent Analogs")
print(top_5_latent)
````

## Encode the Search Pool: Convert all historical days into latent vectors $z_i$.

Let's map every 2D SST field $X_t$ in the GLORYS dataset into a 32-dimensional vector $\mathbf{z}_t \in \mathbb{R}^{32}$. We will extract both $\boldsymbol{\mu}_t$ (the mean representation, which serves as our deterministic feature embedding) and store them in a clean pandas.DataFrame for fast distance queries and downstream metric learning.

````python
import torch
import numpy as np
import pandas as pd

# 1. Set VAE to evaluation mode
vae.eval()

latent_records = []
time_steps = sst_data.time.values

print(f"Encoding {len(time_steps)} historical SST fields into 32D latent space...")

with torch.no_grad():
    for i, t_val in enumerate(time_steps):
        # Extract single 2D slice from dataset: shape (1, 1, 64, 64)
        input_tensor = dataset[i].unsqueeze(0).to(device)
        
        # Pass through encoder to extract latent parameters
        mu, logvar = vae.encode(input_tensor)
        
        # Flatten mean vector (mu) to 1D numpy array
        z_vector = mu.cpu().numpy().flatten()
        
        # Record date and latent coordinates
        record = {'time': t_val}
        for dim in range(vae.latent_dim):
            record[f'z_{dim}'] = z_vector[dim]
            
        latent_records.append(record)

# 2. Build structured DataFrame of Latent Vectors
df_latent = pd.DataFrame(latent_records).set_index('time')
print(f"Encoded Search Pool Shape: {df_latent.shape} (Time Steps × Latent Dims)")
df_latent.head()
````

````python
def find_latent_analogs(target_date, df_latent_pool, top_k=5, exclude_same_day=True):
    """
    Computes Latent RMSD between a target date and all candidate days in the encoded pool.
    """
    target_dt = pd.to_datetime(target_date)
    
    if target_dt not in df_latent_pool.index:
        raise ValueError(f"Target date {target_date} not found in encoded search pool.")
        
    # Extract target 32D vector
    z_target = df_latent_pool.loc[target_dt].values
    
    # Compute Euclidean RMSD across 32 latent dimensions
    # RMSD_z = sqrt( mean( (z_target - z_hist)^2 ) )
    differences = df_latent_pool.values - z_target
    latent_rmsd = np.sqrt(np.mean(differences ** 2, axis=1))
    
    # Create output Series
    scores = pd.Series(latent_rmsd, index=df_latent_pool.index, name="Latent_RMSD")
    
    # Optionally exclude the target day itself
    if exclude_same_day:
        scores = scores.drop(index=target_dt)
        
    return scores.sort_values().head(top_k)

# --- Test Query on July 1, 2004 Target ---
top_5_latent = find_latent_analogs('2004-07-01', df_latent, top_k=5)

print("Top 5 VAE Latent Analogs for 2004-07-01")
for rank, (t_val, dist) in enumerate(top_5_latent.items(), 1):
    date_str = pd.to_datetime(t_val).strftime("%Y-%m-%d")
    print(f"Rank {rank}: Date = {date_str} | Latent RMSD = {dist:.6f}")
````

````python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Reduce 32D latent pool to 2D for visualization
pca = PCA(n_components=2)
z_2d = pca.fit_transform(df_latent.values)

# Get day of year for seasonal color mapping
day_of_year = pd.to_datetime(df_latent.index).dayofyear

plt.figure(figsize=(9, 6))
scatter = plt.scatter(z_2d[:, 0], z_2d[:, 1], c=day_of_year, cmap='twilight', alpha=0.8, edgecolors='none')
cbar = plt.colorbar(scatter)
cbar.set_label('Day of Year (Seasonality)', rotation=270, labelpad=15)

# Highlight Target Date (July 1)
target_idx = df_latent.index.get_loc(pd.to_datetime('2004-07-01'))
plt.scatter(z_2d[target_idx, 0], z_2d[target_idx, 1], color='red', s=120, marker='*', label='Target (July 1)')

plt.title("2D Projection of VAE Latent Space (2004 GLORYS SST)")
plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}% Variance)")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}% Variance)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
````

## Forecast-Guided Metric Learning ($z_t \rightarrow z_{t+30}$)

Let's train a Latent Transition Network $T_\phi$ that maps today's state $z_t$ to the state 30 days in the future $z_{t+30}$.

````python
import torch
from torch.utils.data import Dataset, DataLoader

class LatentTransitionDataset(Dataset):
    def __init__(self, df_latent, horizon=30):
        """
        Creates pairs of latent vectors (z_t, z_{t+horizon}).
        """
        self.horizon = horizon
        self.z_data = torch.tensor(df_latent.values, dtype=torch.float32)
        
        # Valid pairs are from index 0 to len - horizon
        self.n_pairs = len(df_latent) - horizon

    def __len__(self):
        return self.n_pairs

    def __getitem__(self, idx):
        z_t = self.z_data[idx]
        z_t_future = self.z_data[idx + self.horizon]
        return z_t, z_t_future

# Create DataLoader for 30-day ahead transitions
transition_dataset = LatentTransitionDataset(df_latent, horizon=30)
transition_loader = DataLoader(transition_dataset, batch_size=16, shuffle=True)
````

````python
# Define & Train the Transition Network
import torch.nn as nn

class LatentTransitionMLP(nn.Module):
    def __init__(self, latent_dim=32, hidden_dim=64):
        super(LatentTransitionMLP, self).__init__()
        self.mlp = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim),
            nn.GELU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.GELU(),
            nn.Linear(hidden_dim, latent_dim)
        )

    def forward(self, z):
        # Residual step: z_{t+30} = z_t + Δz
        return z + self.mlp(z)

# Initialize
transition_model = LatentTransitionMLP(latent_dim=32, hidden_dim=64).to(device)
opt_trans = torch.optim.Adam(transition_model.parameters(), lr=1e-3)
criterion = nn.MSELoss()

# Training loop
epochs_trans = 450
transition_model.train()

for epoch in range(1, epochs_trans + 1):
    total_loss = 0.0
    for z_t, z_future in transition_loader:
        z_t, z_future = z_t.to(device), z_future.to(device)
        
        opt_trans.zero_grad()
        z_pred = transition_model(z_t)
        loss = criterion(z_pred, z_future)
        
        loss.backward()
        opt_trans.step()
        total_loss += loss.item()
        
    if epoch % 20 == 0 or epoch == 1:
        print(f"Transition Epoch {epoch:03d}/{epochs_trans} | MSE Loss: {total_loss/len(transition_loader):.6f}")
````

````python
transition_model.eval()

# 1. Get current day target vector (e.g., June 1, 2004)
target_start_date = '2004-06-01'
z_june1 = torch.tensor(df_latent.loc[target_start_date].values, dtype=torch.float32).unsqueeze(0).to(device)

# 2. Predict where the latent vector will be in 30 days (July 1, 2004)
with torch.no_grad():
    z_july1_pred = transition_model(z_june1).cpu().numpy().flatten()

# 3. Find analogs in the pool closest to the PREDICTED July 1 vector
diffs = df_latent.values - z_july1_pred
latent_distances = np.sqrt(np.mean(diffs**2, axis=1))

df_forecast_analogs = pd.Series(latent_distances, index=df_latent.index).sort_values()

print(f"30-Day Forecast-Guided Analogs (Predicted state from {target_start_date})")
print(df_forecast_analogs.head(5))
````
